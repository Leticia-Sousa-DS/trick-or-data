# Trick or Data: Data Cleaning

## Objective

This notebook focuses on preparing the Halloween Candy Power Ranking dataset for further analysis.

The process includes data inspection, quality checks, necessary transformations, and validation of the resulting dataset.

## Dataset

- **Source:** FiveThirtyEight
- **Input:** Raw Halloween Candy Power Ranking dataset
- **Output:** Cleaned dataset stored in `data/processed/`
- **Tools:** Python, Pandas, and NumPy

## 1. Data Loading

The raw dataset is retrieved from the original FiveThirtyEight repository and loaded into a Pandas DataFrame.

This step establishes the input data for the cleaning process while preserving the original dataset as the source of truth.

```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/fivethirtyeight/data/master/candy-power-ranking/candy-data.csv"

df = pd.read_csv(url)
```

## 3. Data Quality Checks

This section evaluates the raw dataset for duplicate records, missing values, and invalid values in binary attributes.

```python
# Check duplicate records
print(f"Duplicate records: {df.duplicated().sum()}")
```

```python
# Check missing values
print("\nMissing values per column:")
display(df.isnull().sum())
```

```python
# Check binary attributes
binary_columns = [
    'chocolate', 'fruity', 'caramel', 'peanutyalmondy',
    'nougat', 'crispedricewafer', 'hard', 'bar', 'pluribus'
]

for column in binary_columns:
    print(f"{column}: {sorted(df[column].unique())}")
```

### Data Quality Assessment

The dataset passed the initial quality checks:

- No duplicate records were found.
- No missing values were detected.
- All binary attributes contain only valid values (0 and 1).

No data quality issues were identified during these checks.

## 6. Exporting Cleaned Data

The validated dataset is exported to the processed data directory in CSV format.

Since no cleaning transformations were required, the exported dataset preserves the original records and values.

```python
from pathlib import Path

output_dir = Path('../data/processed')
output_dir.mkdir(parents=True, exist_ok=True)

output_path = output_dir / 'candy_data_cleaned.csv'

df.to_csv(output_path, index=False)

print(f"Dataset exported to: {output_path}")
```
